### 2. Tratamiento de Datos Faltantes (Data Cleaning)

#### 2.1. Imputación de la variable `bathrooms`
Durante el diagnóstico inicial detectamos aproximadamente 3.000 valores nulos en la columna `bathrooms`. Para tratar esta anomalía, definimos la siguiente estrategia:

* **Mecanismo de pérdida (MAR):** Descartamos que los datos falten de forma completamente aleatoria (MCAR). Asumimos un mecanismo *Missing At Random* (MAR), ya que la cantidad de baños esperada está directamente correlacionada con la capacidad máxima de la propiedad (`accommodates`). 
* **Decisión metodológica:** Se descartó la eliminación de filas para no perder el 10% de la muestra. Asimismo, se descartó la imputación global por la media o mediana de toda la columna. En su lugar, aplicamos una **imputación condicional**: reemplazamos los nulos con la mediana de baños del subgrupo de propiedades que comparten la misma capacidad de huéspedes.
* **Efectos sobre la distribución y varianza:** Imputar valores mediante la media o mediana global atenúa artificialmente la variabilidad y las correlaciones porque introduce valores constantes generalizados. Al agrupar por `accommodates` e imputar condicionalmente, logramos rescatar los registros minimizando el impacto sobre la varianza original y preservando la forma de la distribución real de los datos.

#### 2.2. Imputación en Cascada basada en Jerarquías Estructurales
Para tratar los valores nulos en la configuración física de la propiedad (`bathrooms`, `bedrooms`, `beds`), implementamos un método de imputación condicional en cascada que respeta las **relaciones jerárquicas** y evita violar **dependencias lógicas u obvias**. 

El orden de imputación refleja la dependencia estructural real de un alojamiento:
1. **Atributo Raíz:** `accommodates` (capacidad máxima) actúa como la variable independiente principal.
2. **Imputación de Espacios:** Los nulos en `bathrooms` y `bedrooms` se rellenan utilizando la mediana del subgrupo determinado por `accommodates`.
3. **Imputación de Mobiliario:** Los nulos en `beds` se rellenan utilizando la mediana agrupada por `bedrooms`, ya que la cantidad de camas depende físicamente de la disponibilidad de cuartos.
    
Esta técnica secuencial garantiza que las distribuciones imputadas no solo controlen la varianza, sino que mantengan total coherencia con las reglas del dominio inmobiliario.

```python
# Limpiamos los caracteres de formato (corchetes, comillas) y separamos por comas
amenities_limpios = df_copia['amenities'].str.replace(r'\[|\]|"|\'', '', regex=True).str.split(r',\s*')

# Desglosamos las listas y contamos la frecuencia absoluta de cada comodidad individual
conteo_amenities = amenities_limpios.explode().value_counts()

print("--- Top 15 Amenities más frecuentes en la base ---")
print(conteo_amenities.head(15))
```

#### 2.3. Transformación de Atributos Complejos: `amenities`

Para el tratamiento de la variable multivaluada `amenities`, se descartó el enfoque ingenuo de seleccionar las 5 comodidades más frecuentes por conteo absoluto. Dicho método presentaba dos limitaciones severas:
1. **Redundancia y Multicolinealidad:** Atributos como `Dishes and silverware` y `Cooking basics`, `Refrigarator` presentaban una correlación casi directa con `Kitchen`, duplicando información.
2. **Bajo Valor de Negocio:** Atributos triviales de alta frecuencia (como `Hangers` o `Hot water`) no aportan capacidad explicativa sobre la variación de tarifas o la demanda.

**Solución aplicada (One-Hot Encoding por Dominio de Negocio):**
Se seleccionó un conjunto refinado de comodidades estratégicas y no redundantes para transformar en variables binarias ($1/0$):
* `has_kitchen`: Infraestructura para cocinar.
* `has_wifi`: Conectividad wifi.
* `has_ac`: Aire acondicionado.
* `has_bed_linens`: Equipamiento de cama.
* `has_TV`: TV.

```python
import pandas as pd

# 1. Definimos un diccionario con los nombres limpios y sus expresiones regulares (regex)
patrones_amenities = {
    'Wifi / Internet': 'Wifi|Internet',
    'Kitchen (Cocina)': 'Kitchen',
    'TV (Cualquier tipo)': 'TV',
    'Air conditioning (AC)': 'Air conditioning|AC',
    'Hot water': 'Hot water',
    'Bed linens': 'Bed linens',
    'Refrigerator': 'Refrigerator',
    'Pool (Pileta)': 'Pool',
    'Parking (Estacionamiento)': 'Parking',
    'Gym (Gimnasio)': 'Gym',
    'Elevator (Ascensor)': 'Elevator',
    'Balcony / Patio': 'Balcony|Patio|Terrace|Backyard'
}

# 2. Calculamos el porcentaje para cada macro-categoría al vuelo
resultados = {}
for nombre, patron in patrones_amenities.items():
    # .mean() calcula la proporción de Trues (1s). Multiplicado por 100 da el porcentaje.
    porcentaje = df_copia['amenities'].str.contains(patron, case=False, na=False, regex=True).mean() * 100
    resultados[nombre] = porcentaje

# 3. Convertimos los resultados a un DataFrame temporal para ordenarlo y mostrarlo
df_porcentajes = pd.DataFrame(list(resultados.items()), columns=['Amenity_Agrupado', 'Porcentaje (%)'])
df_porcentajes = df_porcentajes.sort_values(by='Porcentaje (%)', ascending=False).reset_index(drop=True)

print("--- Porcentaje Real de Amenities (Agrupado por Regex) ---")
display(df_porcentajes.round(2))
```

#### 2.4. Transformación de `amenities`: Filtro por Varianza y One-Hot Encoding

Al analizar la frecuencia de las comodidades en los alojamientos, evitamos aplicar un *One-Hot Encoding* sobre todas las categorías para no generar un dataset excesivamente esperso. En su lugar, aplicamos un **filtro por varianza**:

1. **Descarte por Varianza Cercana a Cero (*Near-Zero Variance*):** 
   Comodidades como `Wifi` (>99%), `Air conditioning` (>97%), `Kitchen` (>94%) y `TV` (>92%) fueron descartadas. Al estar presentes en casi toda la oferta, se comportan como constantes. Su **entropía es mínima** y no aportan ganancia de información para diferenciar propiedades o justificar precios.
   
2. **Selección de Variables Estratégicas:** 
   Conservamos 6 atributos cuya presencia en la base oscila entre el 25% y el 75%: `Refrigerator`, `Bed linens`, `Elevator`, `Parking`, `Balcony/Patio` y `Pool`. En este rango la **varianza y la entropía son altas**, otorgando gran poder discriminatorio para segmentar el mercado (ej. diferenciar propiedades estándar de las *premium*).

3. **Transformación:**
   Estos 6 atributos clave fueron transformados en **variables dicotómicas (1/0)** mediante *regex*, dejando la base lista para el análisis estadístico sin problemas de multicolinealidad.

```python
# Definimos el diccionario con las comodidades estratégicas seleccionadas y sus patrones de búsqueda (Regex)
amenities_seleccionados = {
    'has_refrigerator': 'Refrigerator',
    'has_bed_linens': 'Bed linens',
    'has_elevator': 'Elevator',
    'has_parking': 'Parking',
    'has_balcony_patio': 'Balcony|Patio|Terrace|Backyard',
    'has_pool': 'Pool'
}

# Generamos las variables binarias (One-Hot Encoding por Dominio)
for col_name, pattern in amenities_seleccionados.items():
    df_copia[col_name] = df_copia['amenities'].str.contains(pattern, case=False, na=False, regex=True).astype(int)
```

### 2. Tratamiento de Datos Faltantes (Data Cleaning)

#### 2.1. Imputación de la variable `bathrooms`
Durante el diagnóstico inicial detectamos aproximadamente 3.000 valores nulos en la columna `bathrooms`. Para tratar esta anomalía, definimos la siguiente estrategia:

* **Mecanismo de pérdida (MAR):** Descartamos que los datos falten de forma completamente aleatoria (MCAR). Asumimos un mecanismo *Missing At Random* (MAR), ya que la cantidad de baños esperada está directamente correlacionada con la capacidad máxima de la propiedad (`accommodates`). 
* **Decisión metodológica:** Se descartó la eliminación de filas para no perder el 10% de la muestra. Asimismo, se descartó la imputación global por la media o mediana de toda la columna. En su lugar, aplicamos una **imputación condicional**: reemplazamos los nulos con la mediana de baños del subgrupo de propiedades que comparten la misma capacidad de huéspedes.
* **Efectos sobre la distribución y varianza:** Como advierte la teoría estadística, imputar valores mediante la media o mediana global atenúa artificialmente la variabilidad y las correlaciones porque introduce valores constantes generalizados. Al agrupar por `accommodates` e imputar condicionalmente, logramos rescatar los registros minimizando el impacto sobre la varianza original y preservando la forma de la distribución real de los datos.

```python
# Obtiene los valores únicos y filtra los de menos de 4 caracteres
valores_unicos = df_copia['property_type'].dropna().unique()
nombres_cortos = [nombre for nombre in valores_unicos if len(str(nombre)) < 10]

print(nombres_cortos)
```
